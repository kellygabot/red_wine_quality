```python
# setup

import pandas as pd
from sklearn.model_selection import train_test_split
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

file_path = '../../dataset/winequality-red.csv'

try:
    data = pd.read_csv(file_path)
    print("Dataset loaded successfully.")
except FileNotFoundError:
    raise FileNotFoundError(
        f"Dataset not found at: {file_path}\n"
        "Upload the CSV file and update the file_path variable"
    )  
    
data = data.dropna() 
data
```

```python
# split the data
X = data.drop(columns=['quality'])
y = data['quality']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=2, stratify=y
)

print(X_train.shape, X_test.shape)
```

```python
# get error baseline model that uses the mean y value in the training set to predict the test set

# 1. Calculate the mean of the target variable from y_train
average_quality = y_train.mean()

# 2. Predict the average value for every row in the test set
baseline_model_test_predictions = [average_quality] * len(y_test)

# 3. Evaluate the baseline MAE
baseline_mae = mean_absolute_error(y_test, baseline_model_test_predictions)
print(f"Baseline MAE: {baseline_mae:.4f}")

# 4. Evaluate the baseline MSE
baseline_mse = mean_squared_error(y_test, baseline_model_test_predictions)
print(f"Baseline MSE: {baseline_mse:.4f}")

# 5. Evaluate the baseline R2
baseline_r2 = r2_score(y_test, baseline_model_test_predictions)
print(f"Baseline R2: {baseline_r2:.4f}")
```
